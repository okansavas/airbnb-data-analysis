```python
# 04_training_airbnb.ipynb - GÜNCELLENMİŞ
# ==========================================================
# 🧠 Airbnb - Model Training
# ==========================================================

import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import joblib
import time
import os

print("🚀 Training gestartet...")

# ========== 0️⃣ Datenüberprüfung ==========
print("🔍 Überprüfe Daten vor dem Training...")

# Daten laden
data_path = os.path.join(os.path.dirname(os.getcwd()), "data", "processed")
X_train = pd.read_csv(os.path.join(data_path, "X_train.csv"))
y_train = pd.read_csv(os.path.join(data_path, "y_train.csv"))
X_test = pd.read_csv(os.path.join(data_path, "X_test.csv"))
y_test = pd.read_csv(os.path.join(data_path, "y_test.csv"))

print(f"✅ Daten geladen: {X_train.shape[0]} Trainingsbeispiele, {X_test.shape[0]} Testbeispiele")

# Veri tiplerini kontrol et
print("X_train Datentypen:")
print(X_train.dtypes.value_counts())

# Kategorik veri var mı kontrol et
categorical_cols = X_train.select_dtypes(include=['object']).columns
if len(categorical_cols) > 0:
    print(f"❌ WARNUNG: Kategorische Daten gefunden: {list(categorical_cols)}")
    # Otomatik düzeltme
    for col in categorical_cols:
        X_train[col] = pd.to_numeric(X_train[col], errors='coerce')
        X_test[col] = pd.to_numeric(X_test[col], errors='coerce')
    X_train = X_train.fillna(0)
    X_test = X_test.fillna(0)
    print("✅ Kategorische Daten automatisch konvertiert")
else:
    print("✅ Alle Daten sind numerisch")

# y-Daten als Array konvertieren
y_train = y_train.values.ravel()
y_test = y_test.values.ravel()

print("✅ Datenüberprüfung abgeschlossen")

# ========== 1️⃣ Modell definieren ==========
print("\n🧠 Initialisiere RandomForestRegressor...")
model = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    n_jobs=-1,
    random_state=42
)

# ========== 2️⃣ Training starten ==========
print("\n🚀 Starte Training...")
start_time = time.time()
model.fit(X_train, y_train)
training_time = time.time() - start_time
print(f"✅ Training abgeschlossen in {training_time:.2f} Sekunden")

# ========== 3️⃣ Modellbewertung ==========
print("\n📊 Bewerte das Modell...")
y_pred = model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)
print(f"📉 RMSE: {rmse:.4f}")
print(f"📈 R²-Score: {r2:.4f}")

# ========== 4️⃣ Modell speichern ==========
models_path = os.path.join(os.path.dirname(os.getcwd()), "models")
os.makedirs(models_path, exist_ok=True)
model_path = os.path.join(models_path, "airbnb_model.pkl")
joblib.dump(model, model_path)
print(f"\n💾 Modell gespeichert unter: {model_path}")

# ========== 5️⃣ Ergebnisübersicht ==========
print("\n🎯 Zusammenfassung:")
print(f"- Trainingszeit: {training_time:.2f} Sekunden")
print(f"- RMSE: {rmse:.4f}")
print(f"- R²: {r2:.4f}")
print("✅ Modelltraining erfolgreich abgeschlossen!")
```
